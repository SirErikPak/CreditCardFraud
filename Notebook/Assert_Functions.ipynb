```python
# 1. Define a standard mapping dictionary for testing
TEST_MAPPING = {
    '1': 'Network A',
    '2': 'Network B',
    '3': 'Network C',
    '4': 400,
    '5': 'Five Guys'
}

CUSTOM_DEFAULT = "No Match Found"

print("--- Running Tests for map_first_digit_to_value ---")
```

```python
# --- Test Group 1: Successful Matches ---
# 1.1 Standard string inputs
assert map_first_digit_to_value("12345", TEST_MAPPING) == 'Network A', "Test 1.1 Failed: Standard '1'"
assert map_first_digit_to_value("2000", TEST_MAPPING) == 'Network B', "Test 1.2 Failed: Standard '2'"
assert map_first_digit_to_value("5999", TEST_MAPPING) == 'Five Guys', "Test 1.3 Failed: Standard '5'"
# 1.4 Test non-string mapped value
assert map_first_digit_to_value("4ABC", TEST_MAPPING) == 400, "Test 1.4 Failed: Integer mapped value"

print("Test Group 1 (Successful Matches) Passed.")
```

```python
# --- Test Group 2: Default Value & No Match ---
# 2.1 Default 'Unknown' should be returned for unmatched digit
assert map_first_digit_to_value("98765", TEST_MAPPING) == "Unknown", "Test 2.1 Failed: Default 'Unknown'"
# 2.2 Custom default should be returned
assert map_first_digit_to_value("65432", TEST_MAPPING, default_value=CUSTOM_DEFAULT) == CUSTOM_DEFAULT, "Test 2.2 Failed: Custom Default"
# 2.3 Input starting with a non-digit character
assert map_first_digit_to_value("Zebra", TEST_MAPPING) == "Unknown", "Test 2.3 Failed: First char not in mapping"

print("Test Group 2 (Default Value & No Match) Passed.")
```

```python
# --- Test Group 3: Input Cleaning & Edge Cases ---
# 3.1 Input as an integer
assert map_first_digit_to_value(3000, TEST_MAPPING) == 'Network C', "Test 3.1 Failed: Integer input"
# 3.2 Input with leading/trailing spaces
assert map_first_digit_to_value(" 1234 ", TEST_MAPPING) == 'Network A', "Test 3.2 Failed: Leading/trailing spaces"
# 3.3 Empty string input
assert map_first_digit_to_value("", TEST_MAPPING) == "Unknown", "Test 3.3 Failed: Empty string"
# 3.4 Input of None (converts to 'None')
assert map_first_digit_to_value(None, TEST_MAPPING) == "Unknown", "Test 3.4 Failed: None input"
# 3.5 Input of just spaces
assert map_first_digit_to_value("    ", TEST_MAPPING) == "Unknown", "Test 3.5 Failed: Only spaces"

print("Test Group 3 (Input Cleaning & Edge Cases) Passed.")
```

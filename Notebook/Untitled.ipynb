```python
def isValid(s):
    if len(s) & 1:
        return False
    
    stack = []
    pairs = {'(': ')', '{': '}', '[': ']'}
    
    for char in s:
        if char in pairs:
            # Opening bracket: push it
            stack.append(char)
        elif not stack or pairs[stack.pop()] != char:
            # Closing bracket: check and pop
            return False
            
    return not stack
```
